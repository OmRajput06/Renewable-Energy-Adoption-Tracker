```python
forecast_df = (
    project_df.groupby("year")["renewables_electricity"]
    .sum()
    .reset_index()
)

forecast_df["year"] = forecast_df["year"].astype(str)

forecast_df["ds"] = pd.to_datetime(
    forecast_df["year"] + "-01-01"
)

forecast_df.rename(
    columns={
        "renewables_electricity": "y"
    },
    inplace=True
)

forecast_df = forecast_df[["ds", "y"]]

forecast_df.head()
```

```python
from prophet import Prophet

model = Prophet()
```

```python
model.fit(forecast_df)
```

```python
future = model.make_future_dataframe(
    periods=6,
    freq="YE"
)

future.tail()
```

```python
forecast = model.predict(future)
```

```python
forecast[["ds", "yhat", "yhat_lower", "yhat_upper"]].tail(10)
```

```python
fig = model.plot(forecast)
```

```python
import plotly.express as px

fig = px.line(
    forecast,
    x="ds",
    y="yhat",
    title="Renewable Electricity Forecast Until 2030",
    labels={
        "ds": "Year",
        "yhat": "Predicted Renewable Electricity (TWh)"
    }
)

fig.show()
```

```python
forecast.to_csv(
    "../data/processed/renewable_energy_forecast.csv",
    index=False
)
```

CO₂ Offset Calculator

```python
def calculate_co2_offset(renewable_twh):
    """
    Calculate estimated CO₂ emissions avoided.

    Parameters:
        renewable_twh (float): Renewable electricity in TWh

    Returns:
        float: Estimated CO₂ offset in million tonnes
    """
    emission_factor = 0.82
    return renewable_twh * emission_factor
```

```python
renewable_energy = 5000  # TWh

co2_saved = calculate_co2_offset(renewable_energy)

print(f"Renewable Electricity: {renewable_energy} TWh")
print(f"Estimated CO₂ Offset: {co2_saved:.2f} million tonnes")
```

```python
forecast["co2_offset"] = forecast["yhat"].apply(calculate_co2_offset)
```

```python
forecast[["ds", "yhat", "co2_offset"]].tail(10)
```

```python
import plotly.express as px

fig = px.line(
    forecast,
    x="ds",
    y="co2_offset",
    title="Estimated CO₂ Emissions Avoided Through Renewable Energy",
    labels={
        "ds": "Year",
        "co2_offset": "CO₂ Offset (Million Tonnes)"
    },
    markers=True
)

fig.show()
```
