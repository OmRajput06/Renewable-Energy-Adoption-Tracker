Data Cleaning 

```python
selected_columns = [
    "country",
    "year",
    "iso_code",
    "population",
    "gdp",
    "renewables_electricity",
    "renewables_share_elec",
    "solar_electricity",
    "wind_electricity",
    "hydro_electricity",
    "electricity_generation",
    "carbon_intensity_elec",
    "greenhouse_gas_emissions"
]

energy_df = df[selected_columns]
```

part 2


```python
project_df = energy_df[
    [
        "country",
        "year",
        "iso_code",
        "renewables_electricity",
        "renewables_share_elec",
        "solar_electricity",
        "wind_electricity"
    ]
]

project_df.head()
```

```python
project_df = project_df[project_df["iso_code"].notna()]
```

```python
project_df.shape
```

```python
project_df.isnull().sum()
```

```python
project_df = project_df.dropna(
    subset=[
        "renewables_electricity",
        "solar_electricity",
        "wind_electricity"
    ]
)
```

```python
project_df.reset_index(drop=True, inplace=True)
```

```python
project_df.info()
project_df.head()
project_df.describe()
```

```python
project_df.to_csv(
    "../data/processed/cleaned_energy_data.csv",
    index=False
)
```

EDA


```python
top_countries = (
    project_df.groupby("country")["renewables_electricity"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

print(top_countries)
```

```python
yearly_renewables = (
    project_df.groupby("year")["renewables_electricity"]
    .sum()
    .reset_index()
)

yearly_renewables.head()
```

```python
import plotly.express as px

fig = px.line(
    yearly_renewables,
    x="year",
    y="renewables_electricity",
    title="Global Renewable Electricity Generation Over Time",
    labels={
        "year": "Year",
        "renewables_electricity": "Renewable Electricity (TWh)"
    },
    markers=True
)

fig.show()
```

```python
# top 10 countries 
top_10 = (
    project_df.groupby("country")["renewables_electricity"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
    .reset_index()
)

top_10
```

```python
import plotly.express as px

fig = px.bar(
    top_10,
    x="renewables_electricity",
    y="country",
    orientation="h",
    title="Top 10 Countries by Renewable Electricity Generation",
    labels={
        "renewables_electricity": "Renewable Electricity (TWh)",
        "country": "Country"
    },
    text="renewables_electricity"
)

fig.update_layout(
    yaxis=dict(categoryorder="total ascending")
)

fig.show()
```

Visualization 3: Solar Electricity Generation Trend

```python
# grp data by yr 
solar_trend = (
    project_df.groupby("year")["solar_electricity"]
    .sum()
    .reset_index()
)

solar_trend.head()
```

```python
import plotly.express as px

fig = px.line(
    solar_trend,
    x="year",
    y="solar_electricity",
    title="Global Solar Electricity Generation Over Time",
    labels={
        "year": "Year",
        "solar_electricity": "Solar Electricity (TWh)"
    },
    markers=True
)

fig.update_traces(
    line=dict(width=3),
    marker=dict(size=6)
)

fig.show()
```
